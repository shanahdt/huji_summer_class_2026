## Three questions for the next eight days


> **What kind of knowledge does corpus analysis produce?**

Every method encodes assumptions about what music is,  
what matters about it, and whose music counts as data.


Pause here. Ask: what do you think counts as data in musicology?
This is the question we'll return to every single day.


> **How can we ask questions that are musicologically meaningful?**

The distance between **frequency** and **meaning** is often a large one.


Just because something appears a lot doesn't mean it matters.
Scale degree 5 is the most common in almost everything — so what?
